```python
import numpy as np
import matplotlib.pyplot as plt

spin='_up'
afm1 = np.loadtxt(f'chg_afm1{spin}.dat')
afm2 = np.loadtxt(f'chg_afm2{spin}.dat')
plt.figure(dpi=100)
plt.plot(afm1[:,0],afm1[:,1]-afm2[:,1],marker='o',label='RDF')
plt.plot(afm1[:,0],afm1[:,2]-afm2[:,2],marker='o',label='cumulative')
# plt.ylim(top=0.05)
plt.ylabel('AFM1-AFM2 (e)')
plt.legend()
```
